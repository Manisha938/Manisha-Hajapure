Merge

```python
emp=pd.DataFrame({'id':[101,102,103,104], 'name':['Aakash','Rohit','Kapil','Govinda'], 'did':[1,2,3,4]})
emp
```

```python
dept=pd.DataFrame({'did':[1,2,3],'dname':['IT','Admin','Sales']})
dept
```

```python
pd.merge(emp,dept,how='inner')
```

```python
pd.merge(emp,dept,how='left')
```

```python
pd.merge(emp,dept,how='right')
```

```python
pd.merge(emp, dept, how='inner', on='dept_id')
```
