###Lineplot
1.Trends over time
2.Time Series analysis
3.upward/downword trend

```python
import numpy as np
```

```python
overs=np.arange(1,21)
team_1=np.random.randint(0,37,20)
team_2=np.random.randint(0,37,20)
```

```python
team_1
```

```python
team1_score=np.cumsum(team_1)
team1_score
```

```python
team2_score=np.cumsum(team_2)
team2_score
```

```python
plt.figure(figsize=(7,3))
plt.style.use('dark_background')
plt.plot(overs,team1_score,marker='o',color='red',label='team1')
plt.plot(overs,team2_score,marker='<',color='yellow',label='team2')
plt.title('Match summary')
plt.xlabel('Overs')
plt.ylabel('Score')
plt.legend()
plt.show()
```
