Bar plot

```python
import matplotlib.pyplot as plt
```

```python
products=['Hot coffee','Cold coffee','Tea','Badam milk']
summer=[30,90,80,50]
```

```python
plt.figure(figsize=(7,3))
plt.bar(products,summer,color='yellow')
plt.title('Sales of product for summer')
plt.xlabel('products')
plt.ylabel('count')
plt.show()
```
