```python
import numpy as np
from sklearn.preprocessing import OrdinalEncoder
```

```python
data = np.array([['low'],['medium'],['high'],['low'],['high']])
data
```

```python
OE = OrdinalEncoder(categories=[['low','medium','high']], dtype=int)
```

```python
OE.fit_transform(data)
```

One Hot Encoding
onehotencoder class

```python
data = np.array([['Apple'],['Orange'],['Banana']])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
OHE= OneHotEncoder(sparse_output=False,dtype=int)
```

```python
OHE.fit_transform(data)
```
